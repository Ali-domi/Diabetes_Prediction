## 1. Introduction

This project focuses on predicting diabetes using supervised machine learning techniques.
The dataset used is the **Pima Indians Diabetes Dataset**, which contains medical diagnostic
measurements for female patients of Pima Indian heritage.

The objective is to apply comprehensive data preprocessing techniques, train multiple
classification algorithms, evaluate their performance using appropriate metrics, and
identify the best-performing model.


```python

# Import required libraries
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, MinMaxScaler
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score, confusion_matrix, roc_curve

from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.svm import SVC
```


## 2. Dataset Loading and Exploration


```python

# Load the dataset
df = pd.read_csv("diabetes.csv")

# Display first rows
df.head()
```

```python

# Dataset information
df.info()
```

```python

# Statistical summary
df.describe()
```


## 3. Data Preprocessing


```python

# Replace zero values with NaN for medically invalid columns
invalid_cols = ['Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI']
df[invalid_cols] = df[invalid_cols].replace(0, np.nan)

# Median imputation
df[invalid_cols] = df[invalid_cols].fillna(df[invalid_cols].median())
```

```python

# Feature and target separation
X = df.drop('Outcome', axis=1)
y = df['Outcome']
```

```python

# Train-test split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)
```

```python

# Feature scaling
scaler_standard = StandardScaler()
scaler_minmax = MinMaxScaler()

X_train_std = scaler_standard.fit_transform(X_train)
X_test_std = scaler_standard.transform(X_test)

X_train_mm = scaler_minmax.fit_transform(X_train)
X_test_mm = scaler_minmax.transform(X_test)
```


## 4. Algorithm Implementation and Evaluation


```python

# Helper function for evaluation
def evaluate_model(model, X_test, y_test):
    y_pred = model.predict(X_test)
    y_prob = model.predict_proba(X_test)[:, 1] if hasattr(model, "predict_proba") else None
    
    return {
        "Accuracy": accuracy_score(y_test, y_pred),
        "Precision": precision_score(y_test, y_pred),
        "Recall": recall_score(y_test, y_pred),
        "F1": f1_score(y_test, y_pred),
        "ROC_AUC": roc_auc_score(y_test, y_prob) if y_prob is not None else None
    }
```

```python

# Logistic Regression
lr = LogisticRegression(max_iter=1000)
lr.fit(X_train_std, y_train)
lr_results = evaluate_model(lr, X_test_std, y_test)

lr_results
```

```python

# K-Nearest Neighbors
knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train_mm, y_train)
knn_results = evaluate_model(knn, X_test_mm, y_test)

knn_results
```

```python

# Naive Bayes
nb_model = GaussianNB()
nb_model.fit(X_train_std, y_train)
nb_results = evaluate_model(nb_model, X_test_std, y_test)

nb_results
```

```python

# Support Vector Machine
svm = SVC(probability=True)
svm.fit(X_train_std, y_train)
svm_results = evaluate_model(svm, X_test_std, y_test)

svm_results
```


## 5. Results Comparison


```python

# Create comparison table
results_df = pd.DataFrame.from_dict({
    "Logistic Regression": lr_results,
    "KNN": knn_results,
    "Naive Bayes": nb_results,
    "SVM": svm_results
}, orient='index')

results_df
```
